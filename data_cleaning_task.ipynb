```python
import pandas as pd

# Working dataset link
url = "https://raw.githubusercontent.com/datablist/sample-csv-files/main/files/customers/customers-100.csv"

# Alternative working mirror link
url2 = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

# Load dataset (using reliable dataset)
df = pd.read_csv(url2)

print("Dataset loaded successfully!")
print(df.head())
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Fill missing numeric values
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    df[col] = df[col].fillna(df[col].mean())

# Fill missing text values
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Standardize text
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.lower().str.strip()

# Save cleaned dataset
df.to_csv("cleaned_data.csv", index=False)

print("Cleaning completed!")
```
